```python

%pip install -q tiktoken

import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4o")

def count_tokens(text):
    return len(encoding.encode(text))

system_prompt = "You are a helpful AI tutor."
user_question = "Explain context windows in simple words."

total_text = system_prompt + "\n" + user_question

print("System prompt tokens:", count_tokens(system_prompt))
print("Question tokens:", count_tokens(user_question))
print("Combined text tokens:", count_tokens(total_text))
```

```python

def build_conversation(messages):
    return "\n".join(messages)

messages = [
    "User: What is AI?",
    "Assistant: AI is artificial intelligence.",
    "User: What is machine learning?",
    "Assistant: Machine learning learns patterns from data.",
    "User: What is an LLM?",
    "Assistant: An LLM is a Large Language Model."
]

for i in range(1, len(messages) + 1):
    conversation = build_conversation(messages[:i])
    print(
        f"Messages: {i}, "
        f"Tokens: {count_tokens(conversation)}"
    )
```

```python

def keep_recent_messages(messages, budget):
    selected = []

    for message in reversed(messages):
        candidate = [message] + selected
        candidate_text = "\n".join(candidate)

        if count_tokens(candidate_text) <= budget:
            selected = candidate
        else:
            break

    return selected


history = [
    "User: I am learning Python.",
    "Assistant: Python is a programming language.",
    "User: Explain functions.",
    "Assistant: Functions are reusable blocks of code.",
    "User: Now explain context windows."
]

recent_history = keep_recent_messages(history, budget=50)

print("Kept messages:")
for message in recent_history:
    print(message)

print("\nTokens used:", count_tokens("\n".join(recent_history)))
```

```python

def reversed_arr(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        arr[left], arr[right] = arr[right], arr[left]

        left += 1
        right -= 1

    return arr


arr = [1, 2, 3, 4, 5]
print(reversed_arr(arr))
```

```python

arr = [1, 2, 2, 3, 1, 2, 4]

freq = {}

for x in arr:
    freq[x] = freq.get(x, 0) + 1

print(freq)
```

```python

arr = [1, 2, 3, 2, 4, 1]

seen = set()

for x in arr:
    if x in seen:
        print(x, "is a duplicate")
    else:
        seen.add(x)
```
